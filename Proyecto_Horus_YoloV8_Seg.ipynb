```python
import yaml

config = {
    'train': '/content/dataset_cartagena/images/train',
    'val': '/content/dataset_cartagena/images/train', # Usamos el mismo para validar por ahora
    'nc': 8, # Número de etiquetas usadas
    'names': ['Mar', 'Arena', 'Edificios', 'Espigon', 'Sombrilla', 'Toldo', 'Multitud_Personas', 'Palmera']
}

with open('/content/dataset_cartagena/data.yaml', 'w') as f:
    yaml.dump(config, f)

print("data.yaml actualizado correctamente!")
```

# Uso del Modelo Entrenado en Video


```python
# Carga tu modelo entrenado
from ultralytics import YOLO
model_path = "/content/drive/MyDrive/Colab Notebooks/PAE_HORUS/Modelo_Para_Pre_Etiquetar/Resultados/Entrenamiento_1_C1_12_Dia_04_y_05/weights/best.pt"
model = YOLO(model_path)

# Ruta a tus imágenes nuevas
nuevas_imagenes_path = "/content/drive/MyDrive/Colab Notebooks/PAE_HORUS/Imagenes_Para_Pre_Etiquetar/C1_12_Dia_06/Parte 2"

# Realiza la predicción y guarda las etiquetas en formato de texto
# save_txt=True es la clave aquí
results = model.predict(
    source=nuevas_imagenes_path,
    save_txt=True,
    conf=0.5, # Ajusta el umbral de confianza según necesites
    imgsz=1240
)
```
