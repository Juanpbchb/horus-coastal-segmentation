# Uso del Modelo Entrenado para Pre Etiquetar:



```python
# Carga tu modelo entrenado
from ultralytics import YOLO
model_path = "/content/drive/MyDrive/Colab Notebooks/PAE_Horus/Modelo_Pre_Etiquetado/Resultados/Entrenamiento_1_C1_12_Dia_04_y_05/weights/best.pt"
model = YOLO(model_path)

# Ruta a tus imágenes nuevas
nuevas_imagenes_path = "/content/drive/MyDrive/Colab Notebooks/PAE_Horus/Modelo_Pre_Etiquetado/Imagenes_para_Pre_Etiquetado/C1_12_Dia_06"

# Realiza la predicción y guarda las etiquetas en formato de texto
# save_txt=True es la clave aquí
results = model.predict(
    source=nuevas_imagenes_path,
    save_txt=True,
    conf=0.5, # Ajusta el umbral de confianza según necesites
    imgsz=1240
)
```
